```python
temp = df[df.origin == "Asia"]
temp1 = temp[["vehicle_weight", "model_year"]].head(n=7)
temp1
```

```python
X = np.array(temp1)
X
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
print(np.linalg.cond(XTX))
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w
```
